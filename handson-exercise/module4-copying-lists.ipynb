```python
list_old = [1, 2, 3]
new_list = list_old
list_old[0] = -5
print('list_old: ', list_old, '\nnew_list: ', new_list)
```

```python
list_old = [1, 2, 3]
list_new = list_old[:] # slicing
list_old[0] = -5
print('list_old: ', list_old, '\nlist_new: ', list_new)
```

```python
list_old = [1,2,3,4]
list_new = list_old[:2]
list_old[0] == -5
print('list_old: ', list_old, '\nlist_new: ', list_new)
```
