```python
def show_truth():
    mysterious_var = 'Surprise!'
    print(mysterious_var)

show_truth()
```

```python
def show_truth():
    mysterious_var = 'New Surprise!'
    print(mysterious_var)

mysterious_var = 'Surprise!'
print(mysterious_var)
show_truth()
print(mysterious_var)
```

```python
def show_truth():
    global mysterious_var
    mysterious_var = 'New Surprise!'
    print(mysterious_var)

mysterious_var = 'Surprise!'
print(mysterious_var)
show_truth()
print(mysterious_var)
```

```python
# list equal to - does not shadow
def show_truth():
    mysterious_var = ['New Surprise']
    print(mysterious_var)

mysterious_var = ['Surprise!']
print(mysterious_var)
show_truth()
print(mysterious_var)
```

```python
# list append - shadows. as list is mutable
def show_truth():
    mysterious_var.append('New Surprise')
    print(mysterious_var)

mysterious_var = ['Surprise!']
print(mysterious_var)
show_truth()
print(mysterious_var)
```
