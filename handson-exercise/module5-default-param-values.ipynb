```python
def print_letter_count(text='This is default string to search', letter='a'):
    counter = 0
    for char in text:
        if char == letter:
            counter += 1
    print(f'Number of {letter} in {text} is {counter}')

print_letter_count('How many letters a are there')
print_letter_count()
print_letter_count('Search here', letter='a')
```
