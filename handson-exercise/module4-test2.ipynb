```python
connections = [
    ('Amsterdam', 'Dublin', 100),
    ('Amsterdam', 'Rome', 140),
    ('Rome', 'Warsaw', 130),
    ('Minsk', 'Prague', 95),
    ('Stockholm', 'Rome', 190),
    ('Copenhagen', 'Paris', 120),
    ('Madrid', 'Rome', 135),
    ('Lisbon', 'Rome', 170),
    ('Dublin', 'Rome', 170),
    ]

counter = 0
flight_time = 0
for connection in connections:
    if connection[1] == 'Rome':
        counter += 1
        flight_time += connection[2]
print(f'{counter} connections lead to Rome with an average flight time of {flight_time/counter} minutes')
```
