```python
# Swapping of 2 numbers 
first = input('Enter your first number: ')
second = input('Enter your second number: ')
print('Before swapping: ', first, second)
first = second
second = first
print('After swapping: ', first, second)
# Does not work this way
```

```python
# Swapping of 2 numbers 
first = input('Enter your first number: ')
second = input('Enter your second number: ')
print('Before swapping: ', first, second)
temporary = first
first = second
second = temporary
print('After swapping: ', first, second)
```

```python
# Swapping of 2 numbers 
first = input('Enter your first number: ')
second = input('Enter your second number: ')
print('Before swapping: ', first, second)
first, second = second, first
print('After swapping: ', first, second)
```

```python
top_cities = ['Chennai', 'Mumbai', 'Pune', 'Agra', 'Hyderabad']
print('Before swapping', top_cities)
top_cities[0], top_cities[3] = top_cities[3], top_cities[0]
print('After swapping', top_cities)
```

```python
top_cities = ['Chennai', 'Mumbai', 'Pune', 'Agra', 'Hyderabad']
print('Before swapping: ', top_cities)
top_cities.sort()
print('After swapping: ', top_cities)
```

```python
random_numbers = [2, 5, 0, -3, 4]
print('Before sorting: ', random_numbers)
random_numbers.sort()
print('After sorting: ', random_numbers)
```

```python
random_numbers = [2, 5, 0, -3, 4]
print('Before sorting: ', random_numbers)
random_numbers.sort()
print('After sorting: ', random_numbers)
random_numbers.sort(reverse=True)
print('After sorting reverse: ', random_numbers)
```

```python
top_cities = ['Chennai', 'Mumbai', 'Agra', 'Hyderabad', 'Pune']
print(sorted(top_cities))
print(top_cities)
```
