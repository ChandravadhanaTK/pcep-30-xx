```python
try:
    value = int(input('Enter a number: '))
    print('The inverse of', value, 'is', 1/value)
except ZeroDivisionError:
    print('You provided 0. Division by 0 is not possible. Sorry!')
except ValueError:
    print('You did not provide a number. So i will not calculate the inverse')
```

```python
try:
    value = int(input('Enter a number: '))
    print('The inverse of', value, 'is', 1/value)
except ZeroDivisionError:
    print('You provided 0. Division by 0 is not possible. Sorry!')
except ValueError:
    print('You did not provide a number. So i will not be able to calculate the inverse.')
except:
    print('Something strange happened here. Sorry!')
```
