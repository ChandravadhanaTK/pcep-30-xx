```python
sample_dict = {
    "mouth": "Mund",
    "finger": "Finger",
    "leg": "Bein",
    "hand": "Hand",
    "face": "Gesicht",
    "nose": "Nase"
}

while True:
    word = input('Enter a word in English or EXIT: ')
    if(word == 'EXIT'):
        print('Bye!')
        break
    elif(word in sample_dict):
        print(f'Translation: {sample_dict[word]}')
    else:
        print('No match!')
```
