```python
def show():
    print(val + 1, end='')
 
val = 5
show()
print(val)
```

```python
my_list = ['aaa', 'bbb', 'ccc']
 
def do(my_list):
   del my_list[1]
   my_list[1] = 'aaa'
 
do(my_list)
print(my_list)
```

```python
def first(x):
    return x ** x
 
def second(x):
    return first(x) * first (x)
 
print(second(2))
```

```python
def func(x, y):
    return x * x
 
print(func(2, 5))
```

```python
def func(a):
   global b
   b = a + a
   return b
 
b = 10
print(func(13))
print(b)
```

```python
def func(x):
  if x == 0:
    return 0
  return x + func(x - 1)
 
print(func(3))
```

```python
def func(x):
  x += 1
  return x
 
x = 5
y = func(x + 2)
print(x, y)
```

```python
def numerical():
  for i in range(10):
    yield i%2
 
for x in numerical():
  print(x, end='-')
```

```python
def my_fun(x):
  try:
    return x / x
  except:
    print('ex')
 
my_fun(1)
my_fun(0)




```
