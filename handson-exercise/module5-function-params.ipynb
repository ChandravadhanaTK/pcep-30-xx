```python
def get_average(input):
    sum = 0.0
    for number in input:
        sum += number
    average = sum/len(input)
    print(average)

get_average([5.0, 3.5, 7.8, 9.9, 10.0])
```

```python
get_average()
```

```python
get_average([1,2,3,4])
```

```python
get_average(5)
```

```python
def print_letter_count(text, letter):
    counter = 0
    for char in text:
        if char == letter:
            counter += 1
    print('Number of', letter, 'is', counter)
```

```python
print_letter_count('Welcome', 'e')
```

```python
print_letter_count('Chandravadhana TK', 'a')
```

```python
print_letter_count('People say nothing is impossible, but i do everything everyday', 'e')
```

```python
print_letter_count(text='Welcome', letter='e')
```

```python
print_letter_count(letter='e', text='welcome')
```
