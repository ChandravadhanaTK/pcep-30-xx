```python
cells = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
for x in cells:
    for y in x:
        print('Element: ', y)
```

```python
table = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
for rows in table:
    for cells in rows:
        print('Element: ', cells)
```

```python
table = [['A1', 'A2', 'A3'], ['B1', 'B2', 'B3']]
for rows in table:
    for cells in rows:
        print(cells, '', end='')
    print()
```
