```python
top_cities = ['Chennai', 'Mumbai', 'Pune', 'Hyderabad', 'Agra']
print(top_cities)
for city in top_cities:
    print('Current City: ', city)
```

```python
top_cities = ['Chennai', 'Mumbai', 'Pune', 'Hyderabad', 'Agra']
print(top_cities)
for city_index in range(len(top_cities)):
    print('Current index: ', city_index, ' | Current City: ', top_cities[city_index])
```
