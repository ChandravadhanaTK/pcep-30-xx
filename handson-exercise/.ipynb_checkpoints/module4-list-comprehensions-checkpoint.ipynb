```python
numbers = [1, 2, 3, 4]
```

```python
numbers = []
for i in range(1, 101):
    numbers.append(i)
print(numbers)
```

```python
numbers = [i for i in range(1, 101)] # List Comprehension
print(numbers)
```

```python
numbers = [i for i in range(1, 101) if i % 3 != 0] # skip 3 multiples
print(numbers)
```

```python
numbers = [i for i in range(3,101, 3)]
print(numbers)
```
