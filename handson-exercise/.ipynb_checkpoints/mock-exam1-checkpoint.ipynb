```python
# What is the expected output of the following code?



data = [1, {}, (2,), (), {3}, [4, 5]]
points = 0
 
for i in range(len(data)):
    if type(data[i]) == list:
        points += 1
    elif type(data[i]) == tuple:
        points += 10
    elif type(data[i]) == set:
        points += 100
    elif type(data[i]) == dict:
        points += 1000
    else:
        points += 10000
 
print(points)
```

```python
# What is the expected output of the following code?



fruits1 = ['Apple', 'Pear', 'Banana']
fruits2 = fruits1
fruits3 = fruits1[:]
 
fruits2[0] = 'Cherry'
fruits3[1] = 'Orange'

print(fruits1)
print(fruits2)
print(fruits3)

res = 0
 
for i in (fruits1, fruits2, fruits3):
    if i[0] == 'Cherry':
        print(i)
        res += 1
    if i[1] == 'Orange':
        print(i)
        res += 10
 
print(res)
```
